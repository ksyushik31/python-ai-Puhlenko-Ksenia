### ⚠️ Важное решение по NaN:
- **НЕ применяем** `fillna(0)` для `elevation`, `inceptionYear`, `coord`
- **Причина:** NaN = «данных нет», 0 = «значение равно нулю»
- **Риск:** Заполнение нулями исказит статистику (среднее, медиану, корреляции)

> 📊 *Проверка типов: `df_parks_info.dtypes`*

````python
# ================================
# 🔍 АНАЛИЗ ЗАПОЛНЕННОСТИ: OPTIONAL-поля в df_parks_info
# ================================

# Список полей для проверки
optional_fields = ['inceptionYear', 'elevation', 'coord', 'continentLabel']

print("📊 Доля заполненных значений в OPTIONAL-полях:\n")

# Словарь для хранения результатов (понадобится для выводов)
fill_rates = {}

# Проходим по каждому полю
for field in optional_fields:
    # Проверяем, есть ли столбец в DataFrame
    if field in df_parks_info.columns:
        # notna() возвращает True для непустых значений
        # .mean() считает долю True (так как True=1, False=0)
        # * 100 — переводим в проценты
        fill_rate = df_parks_info[field].notna().mean() * 100

        # Сохраняем результат в словарь
        fill_rates[field] = fill_rate

        # Выводим в требуемом формате с округлением до 1 знака
        print(f"• {field:15} — заполнено у {fill_rate:5.1f}% записей")
    else:
        print(f"⚠️ Столбец '{field}' не найден в df_parks_info")
        fill_rates[field] = 0  # Записываем 0, если столбца нет
````

````python
# ================================
# 📈 ИНТЕРПРЕТАЦИЯ: Что означают эти цифры?
# ================================

print("\n" + "="*60)
print("🎯 ВЫВОДЫ ДЛЯ ВИЗУАЛИЗАЦИЙ")
print("="*60)

# Проходим по результатам и классифицируем
for field, rate in fill_rates.items():
    if rate > 70:
        status = "✅ ОТЛИЧНО"
        recommendation = "подходит для надёжных графиков и статистики"
    elif rate > 30:
        status = "⚠️ СРЕДНЕ"
        recommendation = "можно использовать, но с оговорками (указывать N)"
    else:
        status = "❌ ПЛОХО"
        recommendation = "визуализации будут ненадёжны, лучше исключить"

    print(f"{field:15} | {rate:5.1f}% | {status:10} | {recommendation}")

# 🎯 Конкретные рекомендации по типам визуализаций
print("\n" + "💡 РЕКОМЕНДАЦИИ ПО ГРАФИКАМ:")
print("-" * 60)

if fill_rates.get('coord', 0) > 70:
    print("🗺️  coord: можно строить интерактивную карту парков")
else:
    print("🗺️  coord: слишком много пропусков → карта будет неполной")

if fill_rates.get('elevation', 0) > 70:
    print("📊 elevation: подойдёт для гистограммы распределения высот")
elif fill_rates.get('elevation', 0) > 30:
    print("📊 elevation: гистограмма возможна, но указать 'N = ...' в подписи")
else:
    print("📊 elevation: лучше не визуализировать, либо показать только примеры")

if fill_rates.get('inceptionYear', 0) > 70:
    print("📅 inceptionYear: можно строить timeline основания парков")
else:
    print("📅 inceptionYear: временной ряд будет фрагментарным")

if fill_rates.get('continentLabel', 0) > 70:
    print("🌍 continentLabel: подойдёт для круговой диаграммы по континентам")
else:
    print("🌍 continentLabel: лучше группировать с другими гео-полями")
````

**Вывод:**  
Благодаря высокой заполненности coord (99.5%) и inceptionYear (97.1%), основные визуализации Задания 3 будут надёжными. Для elevation (50.5%) графики возможны, но с пометкой о размере выборки.

````python
# ================================
# 🔍 ПРОВЕРКА ДУБЛИКАТОВ: по ключу URL
# ================================

# Считаем количество уникальных URL (сколько всего разных парков)
n_unique = df_parks_info['URL'].nunique()

# Считаем общее количество строк в таблице
n_total = len(df_parks_info)

# Выводим сравнение
print(f"📊 Уникальных URL: {n_unique}")
print(f"📊 Всего строк: {n_total}")
print(f"📊 Дубликатов: {n_total - n_unique}")

# Проверяем, есть ли дубликаты
if n_unique < n_total:
    print("\n⚠️ ОБНАРУЖЕНЫ ДУБЛИКАТЫ! Требуется обработка...")
    has_duplicates = True
else:
    print("\n✅ Дубликатов не найдено — данные чистые!")
    has_duplicates = False
````

````python
# ================================
# 📋 АНАЛИЗ ДУБЛИКАТОВ: по каким полям расхождения?
# ================================

if has_duplicates:
    print("\n🔬 Анализируем расхождения в дубликатах:\n")

    # Проверяем, сколько разных значений inceptionYear у одного URL
    if 'inceptionYear' in df_parks_info.columns:
        print("📅 inceptionYear — сколько разных значений на один парк:")
        year_conflicts = df_parks_info.groupby('URL')['inceptionYear'].nunique()
        print(year_conflicts.value_counts().sort_index())
        # value_counts() покажет: у скольких парков 1 год, у скольких 2 года и т.д.

    # Проверяем координаты — если они разные, это серьёзная проблема
    if 'coord' in df_parks_info.columns:
        print("\n🗺️ coord — сколько разных координат на один парк:")
        coord_conflicts = df_parks_info.groupby('URL')['coord'].nunique()
        print(coord_conflicts.value_counts().sort_index())
````

````python
# ================================
# 🧹 ОБРАБОТКА ДУБЛИКАТОВ: оставляем одну строку на парк
# ================================

if has_duplicates:
    # Запоминаем количество строк ДО удаления
    rows_before = len(df_parks_info)

    # Сортируем по inceptionYear (ascending=True — от старого к новому)
    # Это важно: сначала будут самые ранние годы
    if 'inceptionYear' in df_parks_info.columns:
        df_parks_info = df_parks_info.sort_values('inceptionYear', ascending=True)

    # Удаляем дубликаты по URL, оставляя ПЕРВУЮ строку (keep='first')
    # После сортировки первая строка = минимальный год
    df_parks_info = df_parks_info.drop_duplicates(subset='URL', keep='first')

    # Запоминаем количество строк ПОСЛЕ удаления
    rows_after = len(df_parks_info)
    rows_removed = rows_before - rows_after

    # Выводим отчёт
    print("\n✅ Обработка дубликатов завершена:")
    print(f"   Строк до: {rows_before}")
    print(f"   Строк после: {rows_after}")
    print(f"   Удалено: {rows_removed}")

    # Проверяем, что теперь всё чисто
    print(f"\n📊 Проверка: уникальных URL = {df_parks_info['URL'].nunique()}, строк = {len(df_parks_info)}")
````

### 📊 Результаты проверки:

| Метрика | Значение |
|---------|----------|
| Уникальных URL | `{n_unique}` |
| Всего строк | `{n_total}` |
| Дубликатов найдено | `{n_total - n_unique}` |
| Строк удалено | `{rows_removed if has_duplicates else 0}` |
| Строк осталось | `{rows_after if has_duplicates else n_total}` |

### 🔍 Почему возникают дубликаты?
- Данные агрегированы из нескольких источников (Wikidata, API, сайты парков)
- Один парк мог обновляться в разные годы → несколько записей
- Ошибки при парсинге или ручном вводе

### ✅ Наша стратегия обработки:
1. **Группировка** по уникальному ключу `URL`
2. **Сортировка** по `inceptionYear` (от старого к новому)
3. **Оставляем первую строку** = минимальный год основания
4. **Проверка** после удаления: уникальных URL = количество строк

### ⚠️ Почему минимальный год, а не максимальный?
| Поле | Минимальный ✅ | Максимальный ❌ |
|------|---------------|----------------|
| `inceptionYear` | Год **основания** парка | Год **последней записи** (ошибка!) |
| **Аналогия** | Год регистрации компании | Год последнего отчёта |

> 🎯 **Итог:** Каждый парк теперь представлен **одной строкой** — данные готовы к объединению с другими таблицами!

### ✅ Шаг 4: Обработка дубликатов по URL

**Что обнаружили:**  
• Исходно: 4878 строк, из них только 1954 уникальных URL  
• Удалено: 2924 строки-дубли (≈60% данных)

**Почему так много дублей?**  
В Wikidata одно поле может иметь несколько значений для одного объекта:
• Парк мог менять площадь (расширяться) → несколько значений `areaKm2`
• Указаны разные даты: создание, реорганизация, изменение статуса → несколько `inceptionYear`
• Координаты из разных источников → несколько `coord`

**Как обработали:**  
• Отсортировали по `inceptionYear` (возрастание)  
• Оставили первую строку для каждого URL (`drop_duplicates(keep='first')`)  
• Логика: минимальный год = дата первого официального основания парка

**Результат:**  
✅ Теперь каждый парк представлен одной строкой → статистика и визуализации будут корректными

````python
# ================================
# 🗂️ СЛОВАРЬ-ПЕРЕВОДЧИК: Q-код → читаемое название
# ================================

# Словарь для самых частых Q-кодов (можно дополнять по мере анализа)
qcode_to_name = {
    'Q18618819': 'protected area',           # охраняемая территория
    'Q473972': 'national park',              # национальный парк
    'Q151313': 'UNESCO World Heritage Site', # объект Всемирного наследия
    'Q179049': 'nature reserve',             # природный резерват
    'Q107390': 'biosphere reserve',          # биосферный резерват
    'Q14350': 'regional park',               # региональный парк
    'Q121359': 'marine protected area',      # морская охраняемая зона
    'Q1099079': 'state park',                # парк штата
    'Q47053': 'forest park',                 # лесной парк
    'Q850076': 'wildlife refuge',            # заповедник дикой природы
}

print(f"📚 Словарь содержит {len(qcode_to_name)} переводов Q-кодов")
print("📋 Примеры:", dict(list(qcode_to_name.items())[:3]))
````

````python
# ================================
# 🔄 ФУНКЦИЯ: перевод Q-кода в название
# ================================

def get_readable_type(qcode):
    """
    Возвращает читаемое название типа парка.
    Если Q-код есть в словаре — возвращает название.
    Если нет — возвращает исходный Q-код (чтобы не терять данные).
    """
    if pd.isna(qcode):
        return 'unknown'

    # Если уже читаемое название — возвращаем как есть
    if isinstance(qcode, str) and ' ' in qcode and not qcode.startswith('Q'):
        return qcode

    # Если Q-код — ищем в словаре
    qcode_clean = str(qcode).strip()
    return qcode_to_name.get(qcode_clean, qcode_clean)  # если не найдено — вернём Q-код

# Применяем функцию к столбцу с типами
# (используем тот столбец, где у тебя сейчас Q-коды)
source_col = 'instanceType_short' if 'instanceType_short' in df_parks_types.columns else 'instanceType'

if source_col in df_parks_types.columns:
    df_parks_types['type_readable'] = df_parks_types[source_col].apply(get_readable_type)
    print(f"✅ Создан столбец 'type_readable' на основе '{source_col}'")
    print(f"📊 Примеры: {df_parks_types['type_readable'].head(5).tolist()}")
else:
    print(f"⚠️ Столбец '{source_col}' не найден!")
````

````python
# ================================
# 🔄 ИСПРАВЛЕННАЯ ФУНКЦИЯ: перевод Q-кодов
# ================================

# Расширенный словарь (добавь сюда коды из своего топ-20)
qcode_to_name = {
    'Q18618819': 'protected area',
    'Q473972': 'national park',
    'Q34918903': 'heritage park',
    'Q1896949': 'natural monument',
    'Q158454': 'landscape protection area',
    'Q6974775': 'provincial park',
    'Q179049': 'nature reserve',
    'Q22698': 'wilderness area',
    'Q18618822': 'protected landscape',
    'Q839954': 'country park',
    'Q570116': 'forest reserve',
    'Q112128334': 'urban park',
    'Q1316973': 'regional nature park',
    'Q15069452': 'marine park',
    'Q96592721': 'conservation area',
    'Q943017': 'state reserve',
    'Q17343829': 'ecological reserve',
    'Q1191622': 'national nature park',
    'Q1377575': 'natural park',
    'Q3457162': 'biodiversity reserve',
}

def get_readable_type(qcode):
    """Переводит Q-код в читаемое название"""
    if pd.isna(qcode):
        return 'unknown'

    # Приводим к строке и убираем пробелы
    qcode_clean = str(qcode).strip()

    # Если уже читаемое название — возвращаем
    if ' ' in qcode_clean and not qcode_clean.startswith('Q'):
        return qcode_clean

    # Ищем в словаре (без учета регистра)
    for key, value in qcode_to_name.items():
        if key.lower() == qcode_clean.lower():
            return value

    # Если не нашли — возвращаем как есть (можно заменить на 'other')
    return qcode_clean

# Применяем функцию
if source_col in df_parks_types.columns:
    df_parks_types['type_readable'] = df_parks_types[source_col].apply(get_readable_type)
    print("✅ Столбец 'type_readable' обновлён")
````

````python
# ================================
# 📊 ШАГ 6.1: Анализ посетителей по годам
# ================================

print("="*70)
print("📈 АНАЛИЗ ВРЕМЕННЫХ РЯДОВ: df_visitors")
print("="*70)

# --- 1a) Уникальных парков ---
if 'URL' in df_visitors.columns:
    n_parks_vis = df_visitors['URL'].nunique()
    print(f"\n📌 Уникальных парков: {n_parks_vis}")
else:
    # Если столбец называется иначе, ищем похожий
    url_cols = [col for col in df_visitors.columns if 'url' in col.lower() or 'park' in col.lower()]
    if url_cols:
        print(f"⚠️ Столбец 'URL' не найден. Используем '{url_cols[0]}'")
        n_parks_vis = df_visitors[url_cols[0]].nunique()
    else:
        print("⚠️ Не найдено столбца для идентификации парков")
        n_parks_vis = 0

# --- 1b) Распределение по количеству лет на парк ---
if 'year' in df_visitors.columns:

    # groupby — группируем по парку, nunique — считаем уникальные годы
    years_per_park = df_visitors.groupby(df_visitors.columns[0])['year'].nunique()
    years_distribution = years_per_park.value_counts().sort_index()

    print(f"\n📅 Распределение по количеству лет на парк:")
    print(years_distribution)

    # Считаем проценты
    parks_1_year = int(years_distribution.get(1, 0))
    parks_2plus_years = int(years_distribution[years_distribution.index > 1].sum())
    total_parks = parks_1_year + parks_2plus_years

    if total_parks > 0:
        print(f"\n   • Парков с 1 годом данных: {parks_1_year} ({parks_1_year/total_parks*100:.1f}%)")
        print(f"   • Парков с 2+ годами данных: {parks_2plus_years} ({parks_2plus_years/total_parks*100:.1f}%)")
else:
    print("⚠️ Столбец 'year' не найден")
    years_distribution = None
    parks_1_year = 0
    parks_2plus_years = 0
    total_parks = 1

# --- 1c) Топ-20 годов по числу записей ---
if 'year' in df_visitors.columns:

    years_top = df_visitors['year'].value_counts().sort_index().tail(20)

    print(f"\n📅 Топ-20 годов по числу записей:")

    for year, count in years_top.items():
        # Преобразуем 2020.0 → 2020 для красоты
        year_int = int(year) if not pd.isna(year) else year
        print(f"   {year_int:4} | {count:4} записей")
else:
    years_top = None
````

````python
# ================================
# 🛠️ ШАГ 7.1: Функция для унификации URL
# ================================

def clean_url(url):
    """
    Приводит URL к единому формату: извлекает Q-код.
    Пример: 'http://.../Q123' → 'Q123'
    """
    if pd.isna(url):
        return None
    # Приводим к строке, убираем пробелы, делаем заглавные
    url_str = str(url).strip().upper()
    # Если это полный URL Wikidata — берём только последнюю часть (Q-код)
    if 'WIKIDATA.ORG' in url_str:
        return url_str.split('/')[-1]
    # Иначе возвращаем как есть (уже Q-код или другое)
    return url_str

print("✅ Функция clean_url() создана")
print("📝 Пример: clean_url('http://.../Q1993949') → 'Q1993949'")
````

````python
# ================================
# 🧹 ШАГ 7.2: Создаём URL_clean для каждого файла
# ================================

print("\n" + "="*70)
print("🔧 СОЗДАНИЕ СТОЛБЦА URL_clean")
print("="*70)

# --- Обрабатываем df_parks_info ---
print("\n📁 df_parks_info:")
if 'URL' in df_parks_info.columns:
    df_parks_info['URL_clean'] = df_parks_info['URL'].apply(clean_url)
    print(f"✅ Использован столбец 'URL' → создан 'URL_clean'")
elif 'park' in df_parks_info.columns:
    df_parks_info['URL_clean'] = df_parks_info['park'].apply(clean_url)
    print(f"✅ Использован столбец 'park' → создан 'URL_clean'")
else:
    print(f"⚠️ Не найдено 'URL' или 'park'. Столбцы: {df_parks_info.columns.tolist()}")

# --- Обрабатываем df_park_visitors ---
print("\n📁 df_park_visitors:")
if 'URL' in df_park_visitors.columns:
    df_park_visitors['URL_clean'] = df_park_visitors['URL'].apply(clean_url)
    print(f"✅ Использован столбец 'URL' → создан 'URL_clean'")
elif 'park' in df_park_visitors.columns:
    df_park_visitors['URL_clean'] = df_park_visitors['park'].apply(clean_url)
    print(f"✅ Использован столбец 'park' → создан 'URL_clean'")
else:
    print(f"⚠️ Не найдено 'URL' или 'park'. Столбцы: {df_park_visitors.columns.tolist()}")

# --- Обрабатываем df_parks_social ---
print("\n📁 df_parks_social:")
if 'URL' in df_parks_social.columns:
    df_parks_social['URL_clean'] = df_parks_social['URL'].apply(clean_url)
    print(f"✅ Использован столбец 'URL' → создан 'URL_clean'")
elif 'park' in df_parks_social.columns:
    df_parks_social['URL_clean'] = df_parks_social['park'].apply(clean_url)
    print(f"✅ Использован столбец 'park' → создан 'URL_clean'")
else:
    print(f"⚠️ Не найдено 'URL' или 'park'. Столбцы: {df_parks_social.columns.tolist()}")
````

````python
# ================================
# ✅ ШАГ 7.3: Проверка столбцов URL_clean
# ================================

print("\n" + "="*70)
print("🔍 ПРОВЕРКА: столбцы URL_clean")
print("="*70)

for name, df in [('info', df_parks_info),
                 ('visitors', df_park_visitors),
                 ('social', df_parks_social)]:
    if 'URL_clean' in df.columns:
        sample = df['URL_clean'].dropna().unique()[:3]
        print(f"✅ {name:10} | {len(df['URL_clean'].dropna()):5} значений | примеры: {list(sample)}")
    else:
        print(f"❌ {name:10} | столбец 'URL_clean' НЕ создан!")
````

````python
# ================================
# 🔗 ШАГ 7.4: Считаем пересечения по URL_clean
# ================================

print("\n" + "="*70)
print("📊 ПЕРЕСЕЧЕНИЯ ПО URL_clean")
print("="*70)

# Создаём множества только если столбец существует
info_clean = set(df_parks_info['URL_clean'].dropna()) if 'URL_clean' in df_parks_info.columns else set()
visitors_clean = set(df_park_visitors['URL_clean'].dropna()) if 'URL_clean' in df_park_visitors.columns else set()
social_clean = set(df_parks_social['URL_clean'].dropna()) if 'URL_clean' in df_parks_social.columns else set()

# Считаем пересечения
intersect_vis_soc = len(visitors_clean & social_clean)
intersect_vis_info = len(visitors_clean & info_clean)
intersect_soc_info = len(social_clean & info_clean)
intersect_all = len(visitors_clean & social_clean & info_clean)

print(f"\n📈 visitors & social:   {intersect_vis_soc:5} парков")
print(f"📈 visitors & info:     {intersect_vis_info:5} парков")
print(f"📱 social & info:       {intersect_soc_info:5} парков")
print(f"\n🎯 Все три файла:        {intersect_all:5} парков")

# --- Парки только в одном файле ---
print("\n" + "-"*70)
print("📋 Парки только в одном файле:")
only_info = len(info_clean - visitors_clean - social_clean)
only_visitors = len(visitors_clean - info_clean - social_clean)
only_social = len(social_clean - info_clean - visitors_clean)
print(f"   • Только info:     {only_info:5}")
print(f"   • Только visitors: {only_visitors:5}")
print(f"   • Только social:   {only_social:5}")
````

````python
# ================================
# 🔍 ШАГ 9.0: Проверка формата координат
# ================================

import pandas as pd
import numpy as np
import re

print("="*70)
print("🗺️ ШАГ 9: ПАРСИНГ КООРДИНАТ ИЗ coord")
print("="*70)

# Проверяем, загружен ли df_parks_info
try:
    _ = df_parks_info
    print("✅ df_parks_info найден в памяти")
except NameError:
    print("⚠️ df_parks_info не найден. Загружаю из GitHub...")
    url_info = 'https://raw.githubusercontent.com/ksyushik31/python-ai-Puhlenko-Ksenia/main/data/parks_info.csv'
    df_parks_info = pd.read_csv(url_info)
    print("✅ Загрузка завершена!")

# Проверяем, есть ли столбец coord
if 'coord' in df_parks_info.columns:

    print("\n🔍 Примеры значений coord (первые 5 непустых):")
    print("-"*70)

    # .dropna() — убираем пустые значения
    # .head(5) — берём первые 5 строк
    coord_samples = df_parks_info['coord'].dropna().head(5)

    for i, val in enumerate(coord_samples, 1):
        print(f"{i}. '{val}' | тип: {type(val).__name__} | длина: {len(str(val))}")

    # Считаем, сколько всего записей с координатами
    coord_count = df_parks_info['coord'].notna().sum()
    total = len(df_parks_info)

    print(f"\n📊 Статистика:")
    print(f"   • Всего парков: {total}")
    print(f"   • С координатами: {coord_count} ({coord_count/total*100:.1f}%)")
    print(f"   • Без координат: {total - coord_count} ({(total-coord_count)/total*100:.1f}%)")

else:
    print("⚠️ Столбец 'coord' не найден в df_parks_info")
    print(f"📋 Доступные столбцы: {df_parks_info.columns.tolist()}")
````

````python
# ================================
# 🛠️ ШАГ 9.1: Функция для извлечения координат
# ================================

def parse_coord(coord_str):
    """
    Преобразует строку 'Point(lon lat)' → [longitude, latitude]

    Пример:
        Вход:  'Point(148.92 -20.98)'
        Выход: [148.92, -20.98]
    """
    # Проверяем на пустое значение или не строку
    if pd.isna(coord_str) or not isinstance(coord_str, str):
        return pd.Series([None, None])

    # Регулярное выражение для извлечения двух чисел из Point(lon lat)
    # [-\d.]+ — число с возможными минусом, цифрами и точкой
    # \s+ — один или более пробелов между числами
    match = re.search(r'Point\(([-\d.]+)\s+([-\d.]+)\)', coord_str)

    if match:
        # group(1) — первое число (долгота/longitude)
        # group(2) — второе число (широта/latitude)
        lon = float(match.group(1))
        lat = float(match.group(2))
        return pd.Series([lon, lat])

    # Если формат не совпал — возвращаем NaN
    return pd.Series([None, None])

print("✅ Функция parse_coord() создана")

# Тестируем функцию на примерах
print("\n🧪 Тест функции:")
test_values = [
    'Point(148.92 -20.98)',
    'Point(-74.00 40.71)',
    'Point(0.00 0.00)',
    None,
    'Invalid format'
]

for test in test_values:
    result = parse_coord(test)
    print(f"   '{test}' → lon={result[0]}, lat={result[1]}")
````

````python
# ================================
# 🔄 ШАГ 9.2: Применяем функцию к столбцу coord
# ================================

if 'coord' in df_parks_info.columns:

    print("\n" + "="*70)
    print("🔄 ПРИМЕНЕНИЕ ФУНКЦИИ К ДАННЫМ")
    print("="*70)
    print("⏳ Извлечение координат...")

    # .apply(parse_coord) — применяет функцию к каждой строке столбца coord
    # Результат — два новых столбца: longitude и latitude
    df_parks_info[['longitude', 'latitude']] = df_parks_info['coord'].apply(parse_coord)

    print("✅ Координаты извлечены!")

    # Проверка результата
    print(f"\n📊 Результат:")
    print(f"   • Парков с широтой (latitude): {df_parks_info['latitude'].notna().sum()}")
    print(f"   • Парков с долготой (longitude): {df_parks_info['longitude'].notna().sum()}")

    # Процент успешного извлечения
    coord_success = df_parks_info['latitude'].notna().sum() / len(df_parks_info) * 100
    print(f"   • Успешно: {coord_success:.1f}% парков")

else:
    print("⚠️ Столбец 'coord' не найден")
````
