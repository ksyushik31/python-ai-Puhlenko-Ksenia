```python
# Импортируем библиотеку pandas для работы с табличными данными
# as pd — это сокращение, чтобы писать pd вместо pandas (так принято в сообществе)
import pandas as pd

# Импортируем numpy для математических операций (понадобится позже)
# as np — стандартное сокращение
import numpy as np
```

```python
# Сохраняем ссылку на файл в переменную — так код чище и его легче менять
url_parks_info = 'https://raw.githubusercontent.com/ksyushik31/python-ai-Puhlenko-Ksenia/main/data/parks_info.csv'

# Читаем CSV-файл по ссылке и сохраняем результат в переменную df_parks_info
# df = DataFrame (так в pandas называют таблицы)
df_parks_info = pd.read_csv(url_parks_info)

# Выводим размер таблицы: (количество строк, количество столбцов)
print("📊 parks_info.csv shape:", df_parks_info.shape)

# Показываем первые 3 строки, чтобы быстро оценить структуру данных
display(df_parks_info.head(3))
```
